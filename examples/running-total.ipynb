# Running total

A shopping trip, one cell at a time. Cells share one interpreter, so `total` carries over, and each expression's value is shown, like the REPL.

```grizzly
total = 0;
```

`cost` is first assigned inside each block, so it's local and gone after.

```grizzly
{
    cost = 45 * 3;       # bread
    total += cost;
}
total;
```

```grizzly
{
    cost = 18 * 6;       # eggs
    {
        cost -= 18;      # one free: updates this block's cost
    }
    total += cost;
}
total;
```

Functions are values, so adding tax can be one.

```grizzly
with_tax = amount -> amount + amount * 12 / 100;
with_tax(total);
```

A failing cell shows its error, and the session keeps its state.

```grizzly
total / 0;
```

```grizzly
total;
```
